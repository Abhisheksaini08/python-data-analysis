ARRAY RESHAPING

```python
arr=np.arange((12))
print(arr)
```

```python
reshaped=arr.reshape((3,4))
print(reshaped)
```

```python
flattened= reshaped.flatten()
print(flattened)
```

```python
raveled=arr.ravel() #return original aray
print(raveled)
```

```python
raveled=reshaped.ravel()
print(raveled)
```

```python
#transpose
print(reshaped.T)
```
