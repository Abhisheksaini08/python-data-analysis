```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#data structure[restauranyd_id,2021,2022,2023,2024]
sales_data = np.array([
    [1, 150000,  180000,  220000,  250000],  # Paradise Biryani
    [2, 120000,  140000,  160000,  190000],   # Beijing Bites
    [3, 200000,  230000,  260000,  300000],   # Pizza Hub
    [4, 180000,  210000,  240000,  270000],   # Burger Point
    [5, 160000,  185000,  205000,  230000]    # Chai Point
])
```

```python
print("=== Zomato sales analysis===")
print("\n sales data shape",sales_data.shape)
print("\n sample data for 1st 3 rest",sales_data[:3])
```

```python
#total sales per year

print(np.sum(sales_data,axis=0))
yearly_total=(np.sum(sales_data[:,1:],axis=1))
print(yearly_total)
```

```python
#min sales per year
min_sales=print(np.min(sales_data[:,1:],axis=1))
```

```python
#max_Sales per year
max_sales=print(np.max(sales_data[:,1:],axis=0))
```

```python
#average sales per year
avg_Sales_per_year= print(np.mean(sales_data[:,1:],axis=0))

#average sales per res
avg_Sales_per_rest= print(np.mean(sales_data[:,1:],axis=1))
```

```python
#cumulative sum

cumsales=(np.cumsum(sales_data[:,1:],axis=1))
print(cumsales)

import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
plt.plot(np.mean(cumsales, axis=0))

plt.title("Average Cumulative Sales")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
vector1=np.array([1,2,3,4,5])
vector2=np.array([6,7,8,9,10])

print("vector addition",vector1+vector2)

print("vector multiplication",vector1*vector2)

print("dot product",np.dot(vector1,vector2))

angle= np.arccos(np.dot(vector1,vector2)/(np.linalg.norm(vector1)*np.linalg.norm(vector2)))
print(angle)
```
